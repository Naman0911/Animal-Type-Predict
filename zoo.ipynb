# Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
%matplotlib inline
warnings.filterwarnings('ignore')
```

# Column Name

```python
column = ['animal_name' , 'hair' , 'feathers' , 'eggs' , 'milk' , 'airborne' , 'aquatic' , 'predator' , 'toothed' , 'backbone' , 'breathes' , 'venomous' , 'fins' , 'legs' , 'tail' , 'domestic' , 'catsize' , 'type']
column , len(column)
```

# Dataset

```python
df = pd.DataFrame(pd.read_csv('zoo.data'))
df
```

# Getting the By default column
- It was the 1st record but as the dataset was not proper it used the 1st record as an column name

```python
record1 = df.columns 
record1
```

```python
record1 = record1.str.split('.').str[0]
record1
```

```python
df.columns = column
df.columns
```

# Object Columns

```python
object_cols = [features for features in df.columns if df[features].dtype == 'O']
object_cols
```

```python
for cols in object_cols:
    print(df[cols].unique())
```

# Changing type to Int

```python
for cols in object_cols:
    if(cols != 'animal_name'):
        df[cols] = df[cols].astype(int)
```

```python
df.info()
```

# Storing the cleaned Data

```python
df.to_csv('cleaned_zoo_data.csv')
```

# Checking the features categories

```python
for cols in object_cols:
    print(df[cols].unique())
```

# Observation
- Now we are going to drop the animal_name feature because it is not having any genralize pattern or helping in prediction.
- Becuase it is an name and it is unique for all the animals.
- And If we try to keep them then it will require encoding.
- If we do one hot encoding it will form 100 new columns each representing a animal, which will make an sparse matrix and will increase the dimensionality.
- And if we do the label encoding then there are chances of the animal with the same starting letter in name will get the same label , but that is incorrect because they are 2 different animals.
- So doing any type of encoding it will be blunder only.
- So the best option is to drop the particular column.

```python
df.drop('animal_name', axis = 1 , inplace = True)
```

# Now the Data is cleaned and featured it is ready for training.

```python
X = df.iloc[:,:-1]
Y = df.iloc[:,-1]
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split

X_train , X_test , Y_train , Y_test = train_test_split(X , Y , test_size=0.25 , random_state=42 , stratify=Y)
```

# Observation
- As the data is in the binary form we don't need the standardizing

# Model Training

```python
from sklearn.naive_bayes import BernoulliNB

bnb = BernoulliNB()
bnb.fit(X_train , Y_train)
```

# Prediction

```python
Y_pred_test = bnb.predict(X_test)
Y_pred_train = bnb.predict(X_train)
Y_pred_test
```

# Performance Metrics Test Data

```python
from sklearn.metrics import accuracy_score , classification_report , confusion_matrix

print("The Accuracy is =",accuracy_score(Y_test , Y_pred_test))
print(confusion_matrix(Y_test , Y_pred_test))
print(classification_report(Y_test , Y_pred_test))
```

# Performance Metrics on the Train Data

```python
print("The Accuracy is =",accuracy_score(Y_train , Y_pred_train))
print(confusion_matrix(Y_train , Y_pred_train))
print(classification_report(Y_train , Y_pred_train))
```
